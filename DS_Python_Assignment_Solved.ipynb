## 1. String Functions


### 1. 1. Why do `.lower()` and `.upper()` return a new string?

Python strings are immutable. Methods such as `.lower()` and `.upper()` cannot change the original string, so they return a new string.

```python
s = 'Hello World'
print(s.lower())
print(s.upper())
```

### 2. 2. `strip()` results

`s.strip()` returns `'Extras'`. `'***10***'.strip('*')` returns `'10'`.

```python
s = ' Extras \n'
print(repr(s.strip()))
print('***10***'.strip('*'))
```

### 3. 3. `split(',')`

`s.split(',')` returns `['Jane', 'Doe', 'Cars', '5']`, and the result is a list.

```python
s = 'Jane,Doe,Cars,5'
result = s.split(',')
print(result)
print(type(result))
```

### 4. 4. String slicing

`word[1:3]` returns `'el'` and `word[-4:-1]` returns `'ell'`. Negative indices count from the end, with `-1` representing the last character.

```python
word = 'Hello'
print(word[1:3])
print(word[-4:-1])
```

### 5. 5. `int()` vs `float()`

`int(word)` converts a numeric string to an integer, while `float(word)` converts it to a floating-point number. If `word = 'Hi'`, `int(word)` raises a `ValueError` because `'Hi'` is not a valid integer representation.

```python
word = 'Hi'
try:
    print(int(word))
except ValueError as e:
    print(type(e).__name__, e)
```

## 2. Lists in Python


### 6. 6. Iterate directly over a list

A direct `for item in values` loop is simpler and avoids unnecessary index handling.

```python
values = [11, 22, 33]
for item in values:
    print(item)
```

### 7. 7. Lists are mutable

The assignment `values[1] = 95` changes the existing list in place.

```python
values = [11, 22, 33]
values[1] = 95
print(values)
```

### 8. 8. `append()` vs `extend()`

`append(list2)` adds the entire list as one element, while `extend(list2)` adds each element of `list2` individually.

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
appended = list1.copy()
appended.append(list2)
extended = list1.copy()
extended.extend(list2)
print('append:', appended)
print('extend:', extended)
```

### 9. 9. `pop(2)` vs `remove()`

`pop(2)` returns `33` and leaves `[11, 22, 44]`. `remove(value)` removes the first matching value, while `pop(index)` removes and returns the item at an index.

```python
values = [11, 22, 33, 44]
removed = values.pop(2)
print('returned:', removed)
print('list:', values)
```

### 10. 10. `zip()`

Iterating over `zip(list1, list2)` produces pairs `(1, 4)`, `(2, 5)`, and `(3, 6)`.

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
for x, y in zip(list1, list2):
    print(x, y)
```

## 4. Dictionaries in Python


### 16. 16. Keys and values

A key identifies an entry and its value is the associated data. Retrieve a value with `dictionary[key]`.

```python
contact = {'name': 'Alex', 'phone': '555-1234'}
print(contact['phone'])
```

### 17. 17. `dict[key]` vs `dict.get(key)`

`dict[key]` raises `KeyError` for a missing key. `dict.get(key)` returns `None` by default, so `.get()` is safer when the key may be absent.

```python
contact = {'name': 'Alex', 'phone': '555-1234'}
print(contact.get('email'))
try:
    print(contact['email'])
except KeyError as e:
    print(type(e).__name__, e)
```

### 18. 18. Dictionary ordering

In modern Python, dictionaries preserve insertion order. Therefore, newly added key-value pairs appear in insertion order when iterated.

```python
data = {}
data['first'] = 1
data['second'] = 2
print(data)
```

### 19. 19. Modifying while iterating

Changing a dictionary's size during direct iteration can raise `RuntimeError`. A safe pattern is to first build a list of keys to remove, then remove them afterward.

```python
data = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
to_remove = [key for key, value in data.items() if value % 2 == 0]
for key in to_remove:
    data.pop(key)
print(data)
```

### 20. 20. Iterate with `.items()`

Use `for key, value in dictionary.items():` to access each key and value together.

```python
contact = {'name': 'Alex', 'phone': '555-1234'}
for key, value in contact.items():
    print(key, value)
```

## 5. List and Dictionary Comprehension


### 21. 21. Square numbers from 1 to 10

`[i**2 for i in range(1, 11)]` produces `[1, 4, 9, 16, 25, 36, 49, 64, 81, 100]`. The expression `i**2` computes each square, `for i` iterates over values, and `range(1, 11)` supplies 1 through 10.

```python
squares = [i**2 for i in range(1, 11)]
print(squares)
```

### 22. 22. List `[0, 1, 2, 3, 4, 5]`

### 24. 24. Dictionary mapping 1–10 to squares

Use `{i: i**2 for i in range(1, 11)}`.

```python
squares = {i: i**2 for i in range(1, 11)}
print(squares)
```

### 25. 25. Advantage of comprehensions

Comprehensions express collection-building logic concisely and clearly, often using one readable line instead of several lines of loop code.

```python
values = [i**2 for i in range(1, 6)]
print(values)
```

## 6. Sets


### 26. 26. Three defining properties

Sets are unordered collections, contain unique elements, and support set operations such as union and intersection.

```python
values = {1, 2, 3}
print(values)
```

### 27. 27. Adding an existing item

Adding `'blue'` again has no effect because a set stores each unique element only once.

```python
colors = set(['blue', 'green', 'red'])
colors.add('blue')
print(colors)
```

### 28. 28. `.discard()` vs `.remove()`

`.discard()` does nothing when the item is absent, while `.remove()` raises `KeyError` when the item is absent.

```python
colors = {'blue', 'green', 'red'}
colors.discard('yellow')
print(colors)
try:
    colors.remove('yellow')
except KeyError as e:
    print(type(e).__name__, e)
```

### 29. 29. Union and intersection

`set1.union(set2)` contains `{'blue', 'green', 'red', 'yellow'}` and `set1.intersection(set2)` contains `{'blue'}`. Set display order is not guaranteed.

```python
set1 = {'blue', 'green', 'red'}
set2 = {'blue', 'yellow'}
print(set1.union(set2))
print(set1.intersection(set2))
```

### 30. 30. Shorthand operators

Union uses `|` and intersection uses `&`.

```python
set1 = {'blue', 'green', 'red'}
set2 = {'blue', 'yellow'}
print(set1 | set2)
print(set1 & set2)
```
